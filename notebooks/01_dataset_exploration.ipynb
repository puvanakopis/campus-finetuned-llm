## 3. Explore Category Distribution

```python
category_counts = df_raw['category'].value_counts()
print(category_counts)

plt.figure(figsize=(10, 5))
category_counts.plot(kind='bar', color='#2563EB')
plt.title('Distribution of Examples by Domain Category')
plt.xlabel('Category')
plt.ylabel('Number of Examples')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
