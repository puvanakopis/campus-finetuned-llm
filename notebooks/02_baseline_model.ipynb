## 3. Query the Base Model on University-Specific Questions
Notice how the base model either hallucinates generic universities or gives evasive answers because it lacks specific knowledge of Apex National University.

```python
test_queries = [
    "What faculties does Apex National University have?",
    "Which department offers the Software Engineering degree?",
    "Where is the Ada Lovelace Library located and what are its hours?",
    "What is the minimum lecture attendance requirement to sit for exams?"
]

for q in test_queries:
    res = base_assistant.generate_response(user_message=q, max_new_tokens=150, temperature=0.1)
    print(f"\n❓ Query: {q}")
    print(f"🤖 Base Model: {res}")
    print("-" * 50)
```
